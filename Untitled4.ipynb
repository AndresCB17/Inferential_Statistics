```python
import pandas as pd
df = pd.DataFrame()
for i in range(1,101):
  muestras_binomiales = npr.binomial(1000, 0.6, 50)
  col = f"muestras_{i}"
  df[col] = muestras_binomiales

df_media = df.mean().to_frame(name="muestra de las medias")
df_media
```

```python
import seaborn as sns
df = pd.DataFrame()
for i in range(1,101):
  muestras_exponenciales = npr.exponential(scale=0.6, size=50)
  col = f"muestras_{i}"
  df[col] = muestras_exponenciales
df_media = df.mean().to_frame(name="muestra de las medias")
sns.histplot(df_media['muestra de las medias'])
```
