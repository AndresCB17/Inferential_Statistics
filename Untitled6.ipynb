```python
import numpy as np
np.random.seed()
data = np.random.exponential(size=20)
data
```

```python
data.mean()
```

```python
data.var()
```

```python
import pandas as pd
import altair as alt
import numpy as np

# 1. Preparar los datos observados
df_data = pd.DataFrame({'valores': data})

# 2. Generar los datos para la curva teórica (Distribución Exponencial con escala = 1.0)
x_teorico = np.linspace(0, df_data['valores'].max() + 0.5, 100)
y_teorico = np.exp(-x_teorico)
df_teorico = pd.DataFrame({'x': x_teorico, 'y': y_teorico})

# 3. Crear el histograma / densidad observada utilizando transform_density
hist_densidad = alt.Chart(df_data).transform_density(
    'valores',
    as_=['valores', 'density']
).mark_area(opacity=0.4, color='steelblue').encode(
    alt.X('valores:Q', title='Valor'),
    alt.Y('density:Q', title='Densidad')
)

# 4. Crear la línea para la distribución teórica
linea = alt.Chart(df_teorico).mark_line(color='red', size=2).encode(
    x='x:Q',
    y='y:Q'
)

# 5. Combinar y mostrar la gráfica
grafica = (hist_densidad + linea).properties(
    title='Datos Observados (Densidad) vs. Distribución Exponencial Teórica',
    width=500,
    height=300
)

grafica
```

```python
import scipy.stats as stats
import numpy as np

# Parámetros basados en tu variable real 'data'
n = len(data)
alpha = 0.05
grados_libertad = n - 1

# Varianza muestral (corregido de '52' a 's2')
s2 = np.var(data, ddof=1)

# Valores críticos de chi-cuadrado
chi_inferior = stats.chi2.ppf(alpha / 2, grados_libertad)
chi_superior = stats.chi2.ppf(1 - alpha / 2, grados_libertad)

# Intervalo de confianza para la varianza poblacional
limite_inferior = (grados_libertad * s2) / chi_superior
limite_superior = (grados_libertad * s2) / chi_inferior

# Resultados corregidos con llaves para f-strings
print(f"Tamaño de la muestra: {n}")
print(f"Varianza muestral (s2): {s2:.4f}")
print(f"Valor crítico Chi-cuadrado inferior: {chi_inferior:.4f}")
print(f"Valor crítico Chi-cuadrado superior: {chi_superior:.4f}")
print(f"Intervalo de confianza del 95% para la varianza:")
print(f"[{limite_inferior:.4f}, {limite_superior:.4f}]")
```
